```python
import tiktoken
```

```python
import random

_previous_colors = []

def cprint(*args):
    global _previous_colors

    while True:
        # Generate a random background color
        bg = tuple(random.randint(50, 230) for _ in range(3))

        # Calculate perceived luminance
        r, g, b = bg
        luminance = (0.2126 * r + 0.7152 * g + 0.0722 * b)

        # Pick a readable foreground
        fg = (0, 0, 0) if luminance > 140 else (255, 255, 255)

        color = (fg, bg)

        if color not in _previous_colors:
            break

    _previous_colors.append(color)

    if len(_previous_colors) > 2:
        _previous_colors.pop(0)

    fr, fg, fb = color[0]
    br, bg, bb = color[1]

    print(
        f"\033[38;2;{fr};{fg};{fb}m"
        f"\033[48;2;{br};{bg};{bb}m",
        *args,
        "\033[0m",
        sep="",
        end=""
    )
```

```python
def tokenize(text):
    encoding_names = [
        "o200k_harmony",
        "o200k_base",
        "cl100k_base",
        "p50k_base",
        "p50k_edit",
        "r50k_base",
        "gpt2",
        "gpt3",
        "gpt4",
        "gpt4o",
    ]

    available = set(tiktoken.list_encoding_names())

    for name in encoding_names:
        if name not in available:
            print(f"\n--- {name}: not available in this tiktoken version ---")
            continue

        enc = tiktoken.get_encoding(name)
        token_ids = enc.encode(text)

        print(f"\n--- {name} | {len(token_ids)} tokens ---")

        for token_id in token_ids:
            token_bytes = enc.decode_single_token_bytes(token_id)
            piece = token_bytes.decode("utf-8", errors="backslashreplace")

            if not piece.strip():
                cprint(f"[{token_id}]{piece}")
            else:
                cprint(piece)

        print()
```
